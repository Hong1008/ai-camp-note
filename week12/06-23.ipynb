```python
from crewai import Agent, Task, Crew, Process, LLM
from config import GOOGLE_AI_API_KEY
gemma = 'gemini/gemma-4-26b-a4b-it'
```

```python
# LLM 설정
import os

os.environ['GEMINI_API_KEY'] = GOOGLE_AI_API_KEY
llm = LLM(
    model = gemma,
    # base_url = 'http://localhost:1234/v1',
    # api_key=GOOGLE_AI_API_KEY,
    temperature=0.4,
    max_tokens=1000,                    # 생성할 최대 토근 수
    # frequency_penalty=0.1,              # 단어 반복 억제(높을 수록 중복 감소)
    # presence_penalty=0.1,               # 새로운 주제 언급 유도(높을수록 다양성 증가)-> 새로운 이야기로 확장할지 여부
    # top_p=0.9                           # 확률 샘플링 
)

agent = Agent(
    role = 'AI 어시스턴트',                      # 에이전트 역할
    goal = '사용자에게 간단한 환영 인사를 제공',    # 에이전트가 달성할 목표
    backstory = '당신은 친절한 AI 비서로, 언제나 정중하고 도움이 되는 인사를 건넵니다.', # 에이전트 성격/배경
    llm=llm 
)

task = Task(
    description = "사용자에게 환영 인사 한마디를 작성하세요.",      # 에이전트가 수행할 작업 내용
    expected_output = "한 줄의 따뜻한 환영 메시지",               # 기대되는 출력 결과물
    agent = agent                                              # 이 task를 수행할 에이전트
)

# single 에이전트이므로 한개의 팀으로 구성된 crew를 만든거임
crew = Crew(
    agents=[agent],          # 크루에 속한 에이전트 목록
    tasks=[task],           # 크루가 수행할 태스크 목록
    process=Process.sequential, # 순차적 지행 방식으로 태스크
    verbose=True    # 로그 출력
)

result = await crew.kickoff_async()

result
```

```python
researcher = Agent(
    role='AI Researcher',
    goal='멀티 에이전트 시스템의 주요 장범을 3가지 조사하여 설명합니다.',
    backstory='다년간 AI 트렌드를 연구해온 전문가입니다.',
    llm=gemma,
    # tools = [web_search]
)

writer = Agent(
    role='Technical Writer',
    goal="연구 결과를 바탕으로 간략한 결론을 작성합니다.",
    backstory="복잡한 정보를 쉽게 요약하는 작문 전문가입니다.",
    llm=gemma,
)

researcher_task = Task(
    description="멀티 에이전트 시스템의 주요 장점 3가지를 조사하고, 각 장점을 간략히 설명하세요.",
    expected_output="3가지 장점에 대한 간단한 설명 (bullet point 목록)",
    agent=researcher
)

writer_task = Task(
    description="위 조사 결과를 참고하여, 멀티 에이전트 시스템의 장점에 대한 짧은 결론을 작성하세요.",
    expected_output="3~4문장으로 구성된 결론 단락",
    agent=writer,
    context=[researcher_task]   # 이전 태스크의 결과를 활용. preocess=Process.sequential일 때 생략 가능. 생략하는게 좋은 습관은 아님.
)

crew = Crew(
    agents=[researcher, writer],
    tasks=[researcher_task, writer_task],
    process=Process.sequential,             # process는 task 연관되어 있음
    verbose=True
)

crew_output = await crew.kickoff_async()

for idx, task_output in enumerate(crew_output.tasks_output, start=1):
    print(f"\n[Task {idx} Output]\n{task_output}\n")
```
