```python
from google import genai
from config import GOOGLE_AI_API_KEY
from google.genai import types
from pydantic import BaseModel

# model_name = 'gemma-4-26b-a4b-it'
# model_name = 'gemini-3.1-flash-lite'
model_name = 'gemma-4-31b-it'
```

```python
# 클라이언트 초기화 (API 키는 환경변수 GEMINI_API_KEY를 자동으로 참조합니다)

client = genai.Client(api_key=GOOGLE_AI_API_KEY)
response = client.models.generate_content(
    model=model_name,
    contents='안녕하세요! 오늘 점심 메뉴 추천해주세요.',
    config=types.GenerateContentConfig(
        # 여기에 developer 메시지에 해당하는 시스템 지침을 넣습니다.
        system_instruction=(
            "당신은 5성급 호텔의 수석 셰프입니다. "
            "사용자에게 매우 격식 있고 전문적인 어조로 답변해야 하며, "
            "추천하는 메뉴는 한식에 한정합니다."
        ),
        temperature=0.7,
    ),
)
print(response.text)
# 토큰 메타데이터 확인
usage = response.usage_metadata
print(f"입력 토큰 수: {usage.prompt_token_count}")
print(f"출력 토큰 수: {usage.candidates_token_count}")
print(f"전체 토큰 수: {usage.total_token_count}")
```

```python
questions = [
    "리스트와 튜플의 차이가 뭔가요?",
    # "그러면 언제 어떤 걸 쓰는 게 좋나요?",
    # "전체 내용을 표로 정리해주세요.",
]

instructions = (
    "당신은 파이썬 튜터입니다. 초보자에게 간결하고 쉽게 설명합니다."
    "초보자에게 설명하며, 답변은 반드시 핵심만 요약하여 **3문장 이내**로 작성하세요."
    )

# 3. Chat 세션 생성 (system_instruction 적용)
chat = client.chats.create(
    model=model_name,
    config=types.GenerateContentConfig(
        system_instruction=instructions,
        temperature=0.7,
    )
)
# 4. 질문 순차 전송 및 답변 출력 (대화 맥락 유지됨)
for i, question in enumerate(questions, 1):
    print(f"\n--- [질문 {i}] {question} ---")
    
    # send_message를 통해 이전 대화 내용을 품은 상태로 질답을 이어갑니다.
    response = chat.send_message(question)
    
    print(response.text)
    
    # (선택) 토큰 사용량 확인이 필요한 경우
    usage = response.usage_metadata
    print(f"[토큰 사용량 - 입력: {usage.prompt_token_count}, 출력: {usage.candidates_token_count}]")
```

```python
prompt = """
새로 문을 여는 동네 커피숍의 브랜드 소개 문구를 작성해주세요.

조건:
- 4문장으로 작성하세요.
- 따뜻하지만 과장되지 않은 톤으로 쓰세요.
- 공간의 분위기, 커피의 특징, 손님이 느낄 경험을 자연스럽게 담아주세요.
- 제목 없이 본문만 출력하세요."""

for t in [0.0, 0.7, 1.5]:
    response = client.models.generate_content(
        model=model_name,
        contents=prompt,
        config=types.GenerateContentConfig(
            
            temperature=t,
        ),
    )
    print(response.text)

```

```python
response = client.models.generate_content(
    model=model_name,
    contents='AI의 미래에 대해 설명해주세요.',
    config=types.GenerateContentConfig(
        max_output_tokens=100,
        temperature=0.7,
    ),
)
# 1. response.text 대신 candidates 구조 직접 탐색
candidate = response.candidates[0]
# finish_reason이 정상 종료(STOP)가 아닌 MAX_TOKENS 등으로 짤려도 텍스트 추출 가능
if candidate.content and candidate.content.parts:
    truncated_text = candidate.content.parts[0].text
    print(truncated_text)
    print("\n[안내] 문장이 생성 도중 강제로 끊겼습니다.")
else:
    print("생성된 텍스트가 없습니다.")
# 토큰 메타데이터 확인
usage = response.usage_metadata
print(f"입력 토큰 수: {usage.prompt_token_count}")
print(f"출력 토큰 수: {usage.candidates_token_count}")
print(f"전체 토큰 수: {usage.total_token_count}")
```

```python
# Few-shot: 예시를 함께 제공

# user와 model의 턴을 교대로 배치하여 학습시킵니다.
few_shot_contents = [
    # 예시 1
    types.Content(role="user", parts=[types.Part.from_text(text="문장: '정말 최고였어요! 다음에도 또 보고 싶네요.'")]),
    types.Content(role="model", parts=[types.Part.from_text(text="긍정")]),
    
    # 예시 2
    types.Content(role="user", parts=[types.Part.from_text(text="문장: '돈과 시간이 너무 아까운 작품이었습니다.'")]),
    types.Content(role="model", parts=[types.Part.from_text(text="부정")]),
    
    # 예시 3
    types.Content(role="user", parts=[types.Part.from_text(text="문장: '그냥 평범한 액션 영화라 킬링타임용으로 무난합니다.'")]),
    types.Content(role="model", parts=[types.Part.from_text(text="중립")]),
    
    # 실제 분석 요청
    types.Content(role="user", parts=[types.Part.from_text(text="문장: '이 영화를 보고 병을 얻었습니다.'")])
]
response = client.models.generate_content(
    model=model_name,
    contents=few_shot_contents,
    config=types.GenerateContentConfig(
        system_instruction="당신은 입력된 문장의 감성을 오직 '긍정', '부정', '중립' 중 하나로만 판단하는 전문가입니다."
    )
)


print(response.text.strip()) # "부정" 출력
```

```python
# 2. Pydantic 스키마 정의
class TouristPlace(BaseModel):
    name: str
    description: str
class SeoulTouristSpots(BaseModel):
    places: list[TouristPlace]
# 3. 모델 호출 및 스키마 매핑
response = client.models.generate_content(
    model=model_name,
    contents="서울의 주요 관광지 3곳을 알려주세요.",
    config=types.GenerateContentConfig(
        # 페르소나 및 지침 부여
        system_instruction="서울 관광지 정보를 주어진 스키마에 맞게 추출합니다.",
        # 응답을 JSON 형식으로 강제
        response_mime_type="application/json",
        # 파싱할 Pydantic 스키마 등록
        response_schema=SeoulTouristSpots,
        temperature=0.1
    )
)

# 4. response.parsed 속성을 사용해 Pydantic 객체로 직접 사용
spots = response.parsed
print("--- 전체 구조 확인 ---")
print(spots)
print("\n--- 개별 속성 접근 확인 ---")
# 별도의 json.loads 없이 Python 객체로 직접 접근 가능합니다.
for spot in spots.places:
    print(f"관광지명: {spot.name}")
    print(f"설명: {spot.description}\n")
```

```python
class FollowUpTask(BaseModel):
    owner: str
    task: str
    due: str


class MeetingAnalysis(BaseModel):
    summary: str
    decisions: list[str]
    follow_up_tasks: list[FollowUpTask]
    risks: list[str]


meeting_note = """
[B2B 고객사 온보딩 미팅 - 6월 3일 오후 2시]
고객사 박팀장: 다음 달부터 영업팀 35명이 먼저 사용해 보고, 3분기에는 고객지원팀까지 확대하고 싶습니다.
우리 측 이매니저: 초기 계정 생성은 이번 주 금요일까지 완료하겠습니다. 권한 그룹은 영업 관리자, 영업 담당자, 외부 파트너 3단계로 나누겠습니다.
고객사 박팀장: 기존 CRM에서 고객 목록을 가져와야 하는데, 필드명이 조금 다릅니다. 샘플 파일은 내일 오전에 전달하겠습니다.
우리 측 최엔지니어: 샘플 파일을 받으면 매핑 규칙을 확인하고, 다음 주 화요일까지 마이그레이션 가능 여부를 공유하겠습니다.
고객사 박팀장: 교육은 6월 18일 오전으로 잡아주세요. 단, 보안 검토가 늦어지면 일정이 밀릴 수 있습니다.
"""

response = client.models.generate_content(
    model=model_name,
    contents=meeting_note,
    config=types.GenerateContentConfig(
        # 페르소나 및 지침 부여
        system_instruction="회의나 고객 미팅 기록에서 요약, 결정 사항, 담당자별 후속 작업을 분리합니다.",
        # 응답을 JSON 형식으로 강제
        response_mime_type="application/json",
        # 파싱할 Pydantic 스키마 등록
        response_schema=MeetingAnalysis,
        temperature=0.1
    )
)

# 4. response.parsed 속성을 사용해 Pydantic 객체로 직접 사용
analysis = response.parsed
print("--- 전체 구조 확인 ---")
print(analysis)
print("\n--- 개별 속성 접근 확인 ---")
# 1. 회의 요약 (String) 출력
print(f"📍 회의 요약:\n{analysis.summary}\n")
# 2. 결정 사항 (List of String) 출력
print("📌 결정 사항:")
if analysis.decisions:
    for i, decision in enumerate(analysis.decisions, 1):
        print(f"  {i}. {decision}")
else:
    print("  (결정된 사항이 없습니다.)")
print()
# 3. 후속 태스크 목록 (List of FollowUpTask Object) 출력
print("📋 후속 조치 과제 (Follow-up Tasks):")
if analysis.follow_up_tasks:
    for task in analysis.follow_up_tasks:
        print(f"  - [{task.owner}] {task.task} (기한: {task.due})")
else:
    print("  (등록된 후속 과제가 없습니다.)")
print()
# 4. 리스크 요인 (List of String) 출력
print("⚠️ 리스크 및 우려 요인:")
if analysis.risks:
    for risk in analysis.risks:
        print(f"  - {risk}")
else:
    print("  (특별히 식별된 리스크가 없습니다.)")
```

```python
product_brief = """
제품명: 워크플로우 알림봇 Pro
대상: 20~80명 규모의 스타트업 운영팀
핵심 기능: 슬랙 알림 통합, 승인 요청 자동 분배, 지연 업무 리마인드, 주간 처리 현황 요약
강점: 별도 개발 없이 30분 안에 기본 자동화를 설정할 수 있음
프로모션: 이번 달 가입 팀은 14일 무료 체험 제공
"""

system_prompt= """
당신은 최고의 이커머스 마케팅 카피라이터입니다. 
전달받은 제품 정보를 기반으로 다음 3가지 채널에 최적화된 마케팅 문구를 작성하십시오. 
각 채널별 작성 가이드라인은 다음과 같습니다.

1. 이메일 (email):
   - 목적: 뉴스레터 및 할인 이벤트 안내
   - 톤앤매너: 정중하고 부드러우며 상세함 (존댓말 사용)
   - 분량: 3~4문장 내외로 혜택과 특장점을 충분히 서술

2. 앱 푸시 (app_push):
   - 목적: 즉각적인 클릭 유도
   - 톤앤매너: 짧고 강렬함, 호기심 유발 및 혜택 강조
   - 제약 사항: 반드시 30자 이내로 작성하고, 시작과 끝에 적절한 이모지(Emoji) 필수 포함

3. 상세 페이지 (landing_page):
   - 목적: 정보 전달 및 구매 전환
   - 톤앤매너: 전문적이고 신뢰감을 주는 톤
   - 분량: 가독성 좋은 불릿 포인트(Bullet Point) 형태로 핵심 셀링 포인트 3가지를 정리

"""
response = client.models.generate_content(
    model=model_name,
    contents=product_brief,
    config=types.GenerateContentConfig(
        system_instruction=system_prompt,
        temperature=0.1
    )
)

print(response.text)
```
